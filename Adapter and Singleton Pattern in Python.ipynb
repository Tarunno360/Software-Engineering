```python
class Channel:

  _first_channel_created = False

  def __init__(self, name):
    if not Channel._first_channel_created:
      print("First channel must be 'Twitter'. Creating Twitter channel...")
      self.name = "Twitter"
      Channel._first_channel_created = True
    else:
      self.name = name

    self._subscribers = []

  def upload(self, video_title):
    print(f"A new video is uploaded. The title is {video_title}")
    self._notify_subscribers(video_title)

  def _notify_subscribers(self, video_title):
    for subscriber in self._subscribers:
      subscriber.update(video_title)

  def attach(self, subscriber):
    self._subscribers.append(subscriber)

  def detach(self, subscriber):
    self._subscribers.remove(subscriber)



class Subscriber:
  def __init__(self, name):
    self.channels = []
    self.name = name

  def subscribe(self, channel):
    self.channels.append(channel)
    channel.attach(self)

  def unsubscribe(self, channel):
    self.channels.remove(channel)
    channel.detach(self)

  def update(self, video_title):
    print(f"{self.name} received the notification about this video. Video title: {video_title}")



Twitter = Channel('twitter')
My_channel = Channel('my-channel')
sub_1 = Subscriber("sub1")
sub_2 = Subscriber("sub2")

sub_1.subscribe(Twitter)
sub_2.subscribe(My_channel)

Twitter.upload("video1")
My_channel.upload("video2")

sub_2.unsubscribe(My_channel)

Twitter.upload("video3")
My_channel.upload("video4")

print(Twitter is My_channel)
```

```python
#adapter Question 3

class Messenger:
  def sendMessage(self):
    print("Sending message using Messenger")

class Facebook:
  def createFriendsGroup(self):
    print("Creating a Facebook Friends Group")

  def useNewsFeed(self):
    print("Accessing Facebook News Feed")

class Twitch:
  def streamingVideo(self):
    print("Streaming video using Twitch")


class WeChat:
  __instance = None

  def __new__(cls):
    if cls.__instance is None:
      cls.__instance = super(WeChat, cls).__new__(cls)
    return cls.__instance

  def __init__(self):
      self.__messenger = Messenger()
      self.__facebook = Facebook()
      self.__twitch = Twitch()

  def sendMessage(self):
    self.__messenger.sendMessage()

  def createFriendsGroup(self):
    self.__facebook.createFriendsGroup()

  def useNewsFeed(self):
    self.__facebook.useNewsFeed()

  def streamingVideo(self):
    self.__twitch.streamingVideo()
weChat1 = WeChat()
weChat2 = WeChat()


print(weChat1 is weChat2)

weChat1.sendMessage()
weChat2.createFriendsGroup()
weChat2.useNewsFeed()
weChat1.streamingVideo()


print(weChat1)
print(weChat2)
```
